# Orthogonal Projection onto Span

Chapter 1: Vector Space and Inner Product · *Mathematics for AI and Machine Learning* by Fuheng Wu

Project a vector onto a one-dimensional subspace and inspect residual orthogonality.

Companion code for the [interactive demo](https://math4ai.org/companion/chapter-1/projection-span/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch

x = torch.tensor([2.0, 1.0])          # vector to project
a = torch.tensor([1.0, 0.5])          # direction spanning the 1D subspace

# Unit direction â = a / ||a||
a_hat = a / torch.linalg.norm(a)

# Orthogonal projection onto span{a}: proj = (x·â) â
proj = torch.dot(x, a_hat) * a_hat

# Residual is the component orthogonal to a
residual = x - proj

print("projection:", proj.tolist())
print("residual:", residual.tolist())
# Should be ≈ 0: residual ⊥ a
print("orthogonal check ⟨residual, a⟩ =", torch.dot(residual, a).item())
```
