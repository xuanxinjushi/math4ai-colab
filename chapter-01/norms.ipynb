```python
import torch

# Lp unit ball boundary: { x : ||x||_p = 1 }
p = 2.0
theta = torch.linspace(0, 2 * torch.pi, 400)

# For p = 2 the boundary is the unit circle (||x||_2 = 1)
x = torch.cos(theta)
y = torch.sin(theta)
points = torch.stack([x, y], dim=1)

# Check every sampled point lies on the ball (norm should be 1)
norms = torch.linalg.vector_norm(points, ord=p, dim=1)
print("max norm on curve:", norms.max().item())
```
