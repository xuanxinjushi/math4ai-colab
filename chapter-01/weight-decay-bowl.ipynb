Companion code for the [interactive demo](https://math4ai.org/companion/chapter-1/weight-decay-bowl/) on math4ai.org (sign-in required). Run the cell with Shift+Enter; PyTorch is preinstalled on Colab. Change the values at the top and run it again to explore.


```python
import torch
import torch.nn as nn

model = nn.Linear(10, 5)

# weight_decay=λ adds (λ/2)||W||² to the loss — same L2 penalty as in the demo bowl
opt = torch.optim.SGD(model.parameters(), lr=0.01, weight_decay=1e-2)

for _ in range(50):
    x = torch.randn(32, 10)
    y = torch.randn(32, 5)
    loss = model(x).pow(2).mean()         # task loss only; decay is in the optimizer
    opt.zero_grad()
    loss.backward()                       # gradients include λ W from weight decay
    opt.step()

print("||W|| after decay:", model.weight.norm().item())
```
